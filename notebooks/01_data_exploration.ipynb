بررسی محتوای ستون ها 
برای اینکه بتونیم ببینیم چه شکل هایی از مثلا اسم شرکت در ستون شرکت داریم 

```python
for file in files:
    df = pd.read_excel(file)
    print(f"file name: {file.name}")
    columns = df.columns
    print("columns:")
    for column in columns: 
        print(f"\nColumn: {column}")
        print(df[column].unique())
    print("________")
```

بررسی مقادیر خالی در ستون ها

```python
for file in files:
    df = pd.read_excel(file)
    print(f"\nfile name: {file.name}")
    columns = df.columns
    print(f"\ncolumns:")
    for column in columns: 
        print(f"Column: {column}")
        print(df[column].isnull().sum())
    print("________")
```

بررسی مقادیر غیر عادی 
برای تشخیص مبلغ عجیب یا منفی 

```python
for file in files:
    df = pd.read_excel(file)
    print(f"file name: {file.name}")
    columns = df.columns
    print("columns:")
    for column in columns: 
        print(f"\nColumn: {column}")
        print(df[column].describe())
        print(f"\n*****")
    print("________")
```
